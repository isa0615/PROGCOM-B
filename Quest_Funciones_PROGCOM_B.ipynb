FLORISTERÍA RUFO


```python
import random
```

```python
def crear_inventario():
    flores = ["tulipán", "lirio", "rosa", "dahlia", "margarita", "girasol", "clavel", "geranio"]
    seleccion = random.sample(flores, 4)

    inventario = {
        flor: {
            "precio": random.randint(2000, 6000),
            "cantidad": random.randint(10, 50),
            "vendidas": 0
        }
        for flor in seleccion
    }
    return inventario
```

```python
def mostrar_menu(inventario):
    print("\nMENÚ DE FLORES DISPONIBLES")
    for flor, datos in inventario.items():
        print(f"{flor.capitalize():<12} | Precio: ${datos['precio']:<5} | Cantidad: {datos['cantidad']}")
    print("--------------------------------------------------")
```

```python
def realizar_venta(inventario, flor, cantidad, total_ganado):
    if flor not in inventario:
        print(f"La flor '{flor}' no está disponible en la tienda.")
        return total_ganado

    datos = inventario[flor]

    if cantidad > datos["cantidad"]:
        print(f"Solo quedan {datos['cantidad']} unidades de {flor}. No se puede vender esa cantidad.")
        return total_ganado


    datos["cantidad"] -= cantidad
    datos["vendidas"] += cantidad
    venta_total = datos["precio"] * cantidad
    total_ganado += venta_total

    print(f"Venta realizada: {cantidad} {flor}(es) por un total de ${venta_total}")
    if datos["cantidad"] == 0:
        print(f"La flor {flor} se agotó y fue retirada del inventario.")
        del inventario[flor]

    return total_ganado
```

```python
def cerrar_tienda(inventario_inicial, inventario_final, total_ganado):
    print("\nCIERRE DE TIENDA ")
    flores_vendidas = {}

    # Calcular cuántas flores se vendieron de cada tipo
    for flor, datos in inventario_inicial.items():
        cantidad_final = inventario_final.get(flor, {"cantidad": 0})["cantidad"]
        vendidas = datos["cantidad"] - cantidad_final
        flores_vendidas[flor] = vendidas

    # Determinar flores más y menos vendidas
    mas_vendida = max(flores_vendidas, key=flores_vendidas.get)
    menos_vendida = min(flores_vendidas, key=flores_vendidas.get)

    print(f" Flor más vendida: {mas_vendida.capitalize()} ({flores_vendidas[mas_vendida]} unidad(es))")
    print(f" Flor menos vendida: {menos_vendida.capitalize()} ({flores_vendidas[menos_vendida]} unidad(es))")
    print(f" Total recaudado en el día: ${total_ganado}")
    print("--------------------------------------------------")
    print("Gracias por usar el sistema de Floristería RUFITO ")
```

```python
def main():
    print("Bienvenido a la Floristería Rufito\n")

    inventario = crear_inventario()
    inventario_inicial = {k: v.copy() for k, v in inventario.items()}
    total_ganado = 0
    opcion = ""

    while opcion != "3":
        print("\nMENÚ PRINCIPAL")
        print("1. Ver flores disponibles")
        print("2. Vender flor")
        print("3. Cerrar tienda")

        opcion = input("Seleccione una opción: ")

        if opcion == "1":
            mostrar_menu(inventario)

        elif opcion == "2":
            flor = input("Ingrese el nombre de la flor: ").lower()
            cantidad = int(input("Cantidad a vender: "))
            total_ganado = realizar_venta(inventario, flor, cantidad, total_ganado)

        elif opcion == "3":
            cerrar_tienda(inventario_inicial, inventario, total_ganado)
            print("Hasta pronto 👋")
        else:
            print("Opción inválida, intente de nuevo.")
```

```python
if __name__ == "__main__":
    main()
```
